```python
#Palindrome checker. Function that takes a string and returns True if it reads the same forwards and backwards. Handle uppercase/spaces.

def palindrome(text):
    reformatted = text.lower().replace(" ","")
    return reformatted == reformatted[::-1]
    
text = input("please write a text that you want to check:")
print(palindrome(text))


```

```python
#Word counter. Read a text file (find any free book on Project Gutenberg), count how many times each word appears, print the top 10 most common.


with open(r"C:\Users\sethl\OneDrive\Documents\AI Engineering Work\Python for Data Science, AI & Development (1st course)\MobyDickexample.txt", "r", encoding="utf-8") as f:
    text = f.read()
    
    class wordcounter:
        def __init__(self,text):
            self.text = text
        def counter(self):
            lower = self.text.lower()
            splitwords = lower.split(" ")
            dictionary = {}
            
            for word in splitwords:
                if word in dictionary:
                    dictionary[word] = dictionary[word] + 1
                else:
                    dictionary[word] = 1
            return(dictionary)
                
     
WC = wordcounter(text)
result = WC.counter()
top10 = sorted(result.items(), key = lambda item:item[1], reverse = True) [:10]
print(top10)



    
```

```python
# Number guessing game. Computer picks a random number 1-100, user guesses, program says "higher" or "lower" until they get it.
import random

while True:
    selected = random.randint(1, 101)
    while True:
        userguess = input("please input a number you would like to guess:")
        if int(userguess) == selected:
            print("wow you've got it!")
            break
        elif int(userguess) < selected:   
            print("try go higher") 
        elif int(userguess) > selected:
            print("try go lower")
    playagain = input("do you want to play again?")
    if playagain == "no":
        break
    





```
